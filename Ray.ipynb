```python
# -*- coding: utf-8 -*-
"""
================================================================================
SISTEMA DE IDENTIFICACIÓN DE BILLETES PERUANOS FALSIFICADOS
================================================================================
Arquitectura Híbrida CNN-RNN con Procesamiento Digital de Imágenes (PDI)

"""
# =============================================================================
# CELDA 1: INSTALACIÓN Y CONFIGURACIÓN INICIAL
# =============================================================================
# Ejecutar esta celda primero y esperar a que termine

print("="*70)
print(" SISTEMA DE IDENTIFICACIÓN DE BILLETES PERUANOS")
print("="*70)
print("\n Configurando entorno...")

import os
import sys
import random
import warnings
import time
from datetime import datetime

# Configuración para evitar warnings molestos
warnings.filterwarnings('ignore')
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'

import numpy as np
print(f" NumPy: {np.__version__}")

import cv2
print(f" OpenCV: {cv2.__version__}")

import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.patches import Rectangle
plt.style.use('default')
print(" Matplotlib configurado")

import tensorflow as tf
tf.get_logger().setLevel('ERROR')
print(f" TensorFlow: {tf.__version__}")

# Verificar GPU
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    print(f" GPU detectada: {gpus[0].name}")
    for gpu in gpus:
        tf.config.experimental.set_memory_growth(gpu, True)
else:
    print(" No se detectó GPU - El entrenamiento será más lento")

from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import (Conv2D, MaxPooling2D, Dense, Dropout, Flatten,
                                     BatchNormalization, LSTM, GRU, Bidirectional,
                                     Reshape, GlobalAveragePooling2D, Input,
                                     TimeDistributed, Concatenate, Add)
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.utils import to_categorical

from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
from sklearn.model_selection import train_test_split
import pandas as pd

# Semillas para reproducibilidad
np.random.seed(42)
tf.random.set_seed(42)
random.seed(42)

print("\n Todas las librerías cargadas correctamente")
print("="*70)

# =============================================================================
# CELDA 2: DESCARGA DEL DATASET
# =============================================================================

print("\n" + "="*70)
print(" DESCARGANDO DATASET DE BILLETES PERUANOS")
print("="*70)

# Clonar repositorio si no existe
if not os.path.exists('/content/Dataset-Billetes-Peruanos'):
    print(" Descargando dataset desde GitHub...")
    os.system('git clone -q https://github.com/NicolasPCS/Dataset-Billetes-Peruanos.git')
    print(" Dataset descargado")
else:
    print(" Dataset ya existe en el sistema")

# Rutas del dataset
DATASET_PATH = '/content/Dataset-Billetes-Peruanos'
TRAIN_PATH = os.path.join(DATASET_PATH, 'training_set')
VAL_PATH = os.path.join(DATASET_PATH, 'validation_set')
TEST_PATH = os.path.join(DATASET_PATH, 'testing_set')

# Obtener categorías
CATEGORIAS = sorted(os.listdir(TRAIN_PATH))
NUM_CLASES = len(CATEGORIAS)

# Contar imágenes
def contar_imagenes(path):
    total = 0
    for cat in CATEGORIAS:
        cat_path = os.path.join(path, cat)
        if os.path.exists(cat_path):
            total += len([f for f in os.listdir(cat_path) if f.endswith(('.jpg', '.jpeg', '.png'))])
    return total

n_train = contar_imagenes(TRAIN_PATH)
n_val = contar_imagenes(VAL_PATH)
n_test = contar_imagenes(TEST_PATH)

print(f"\n ESTADÍSTICAS DEL DATASET:")
print(f"    Categorías: {NUM_CLASES}")
print(f"    Training:   {n_train:,} imágenes")
print(f"    Validation: {n_val:,} imágenes")
print(f"    Testing:    {n_test:,} imágenes")
print(f"    TOTAL:      {n_train + n_val + n_test:,} imágenes")

print(f"\n CATEGORÍAS DISPONIBLES:")
for i, cat in enumerate(CATEGORIAS, 1):
    cat_path = os.path.join(TRAIN_PATH, cat)
    n = len(os.listdir(cat_path)) if os.path.exists(cat_path) else 0
    print(f"   {i:2d}. {cat}: {n} imágenes")

print("="*70)

# =============================================================================
# CELDA 3: INFORMACIÓN DE BILLETES PERUANOS (BCRP)
# =============================================================================

print("\n" + "="*70)
print(" INFORMACIÓN OFICIAL DE BILLETES PERUANOS (BCRP)")
print("="*70)

INFO_BILLETES = {
    '10': {
        'nuevo': {
            'denominacion': 'S/ 10.00',
            'serie': 'Nueva Familia (2021)',
            'personaje': 'Chabuca Granda',
            'profesion': 'Compositora y cantante criolla (1920-1983)',
            'reverso': 'Vicuña y Flor de Amancaes',
            'color_dominante': 'Verde',
            'color_rgb': (74, 159, 74),
            'dimensiones_mm': '140 x 65',
            'elementos_seguridad': [
                'Marca de agua multitonal (rostro de Chabuca Granda)',
                'Hilo de seguridad (cambia de violeta a bronce)',
                'Tinta OVI (cambia de dorado a verde)',
                'Microimpresiones "BCRP" y "DIEZ"',
                'Registro perfecto (flor visible a trasluz)',
                'Imagen latente con denominación "10"',
                'Relieve táctil en denominación',
                'Fibrillas fluorescentes UV (rojas, azules, verdes)'
            ]
        },
        'antiguo': {
            'denominacion': 'S/ 10.00',
            'serie': 'Serie Anterior (2009)',
            'personaje': 'José Abelardo Quiñones',
            'profesion': 'Héroe de la Aviación Peruana (1914-1943)',
            'reverso': 'Macchu Picchu',
            'color_dominante': 'Azul verdoso',
            'color_rgb': (70, 130, 150)
        }
    },
    '20': {
        'nuevo': {
            'denominacion': 'S/ 20.00',
            'serie': 'Nueva Familia (2022)',
            'personaje': 'José María Arguedas',
            'profesion': 'Escritor y antropólogo (1911-1969)',
            'reverso': 'Cóndor Andino y Flor de la Cantuta',
            'color_dominante': 'Amarillo Ocre',
            'color_rgb': (204, 153, 51),
            'dimensiones_mm': '140 x 65',
            'elementos_seguridad': [
                'Marca de agua con rostro de Arguedas',
                'Hilo de seguridad (cambia de verde a azul)',
                'Tinta OVI Spark Live',
                'Microimpresiones "BCRP" y "VEINTE"',
                'Registro perfecto',
                'Imagen latente "20"'
            ]
        },
        'antiguo': {
            'denominacion': 'S/ 20.00',
            'serie': 'Serie Anterior (2009)',
            'personaje': 'Raúl Porras Barrenechea',
            'profesion': 'Historiador y diplomático (1897-1960)',
            'reverso': 'Complejo Arqueológico Chan Chan',
            'color_dominante': 'Marrón anaranjado',
            'color_rgb': (180, 120, 60)
        }
    },
    '50': {
        'nuevo': {
            'denominacion': 'S/ 50.00',
            'serie': 'Nueva Familia (2022)',
            'personaje': 'María Rostworowski',
            'profesion': 'Historiadora e investigadora (1915-2016)',
            'reverso': 'Jaguar y Puya Raimondi',
            'color_dominante': 'Rojo rosáceo',
            'color_rgb': (178, 34, 34),
            'dimensiones_mm': '140 x 65',
            'elementos_seguridad': [
                'Marca de agua con rostro de Rostworowski',
                'Hilo de seguridad marrón con movimiento',
                'Tinta OVI que cambia de color',
                'Microimpresiones "BCRP" y "CINCUENTA"',
                'Registro perfecto',
                'Número oculto "50"'
            ]
        },
        'antiguo': {
            'denominacion': 'S/ 50.00',
            'serie': 'Serie Anterior (2009)',
            'personaje': 'Abraham Valdelomar',
            'profesion': 'Escritor y periodista (1888-1919)',
            'reverso': 'Laguna y Oasis de Huacachina',
            'color_dominante': 'Gris violáceo',
            'color_rgb': (120, 100, 140)
        }
    },
    '100': {
        'nuevo': {
            'denominacion': 'S/ 100.00',
            'serie': 'Nueva Familia (2021)',
            'personaje': 'Pedro Paulet Mostajo',
            'profesion': 'Científico espacial (1874-1945)',
            'reverso': 'Colibrí Cola de Espátula y Orquídea',
            'color_dominante': 'Azul acero',
            'color_rgb': (70, 130, 180),
            'dimensiones_mm': '140 x 65',
            'elementos_seguridad': [
                'Marca de agua multitonal con rostro de Paulet',
                'Hilo de seguridad (cambia de fucsia a verde)',
                'Tinta OVI Spark Live (efecto 3D)',
                'Microimpresiones "BCRP" y "CIEN"',
                'Registro perfecto',
                'Imagen latente "100"',
                'Elementos magnéticos avanzados',
                'Tinta infrarroja'
            ]
        },
        'antiguo': {
            'denominacion': 'S/ 100.00',
            'serie': 'Serie Anterior (2009)',
            'personaje': 'Jorge Basadre Grohmann',
            'profesion': 'Historiador (1903-1980)',
            'reverso': 'Biblioteca Nacional del Perú',
            'color_dominante': 'Rojo ladrillo',
            'color_rgb': (170, 74, 68)
        }
    }
}

# Mostrar información
for denom, tipos in INFO_BILLETES.items():
    print(f"\n BILLETE DE S/ {denom}")
    for tipo, info in tipos.items():
        print(f"    {tipo.upper()}: {info['personaje']} - {info.get('profesion', 'N/A')}")

print("\n" + "="*70)

# =============================================================================
# CELDA 4: CLASE PDI - PROCESAMIENTO DIGITAL DE IMÁGENES (COMPLETO)
# =============================================================================

print("\n" + "="*70)
print("🔬 INICIALIZANDO MÓDULO DE PROCESAMIENTO DIGITAL DE IMÁGENES (PDI)")
print("="*70)

class PDI:
    """
    Clase completa de Procesamiento Digital de Imágenes para análisis de billetes.
    Implementa técnicas específicas para detección de falsificaciones.
    """

    # =========================================================================
    # MÉTODOS AUXILIARES
    # =========================================================================

    @staticmethod
    def asegurar_uint8(img):
        """Convierte imagen a uint8 de forma segura"""
        if img.dtype != np.uint8:
            img = np.clip(img, 0, 255).astype(np.uint8)
        return img

    @staticmethod
    def asegurar_rgb(img):
        """Convierte imagen a RGB si es necesario"""
        if len(img.shape) == 2:
            return cv2.cvtColor(img, cv2.COLOR_GRAY2RGB)
        return img

    @staticmethod
    def normalizar(img):
        """Normaliza imagen al rango 0-255"""
        if img.max() > img.min():
            return ((img - img.min()) / (img.max() - img.min()) * 255).astype(np.uint8)
        return np.zeros_like(img, dtype=np.uint8)

    # =========================================================================
    # CONVERSIONES DE ESPACIO DE COLOR
    # =========================================================================

    @staticmethod
    def a_gris(img):
        """Convierte a escala de grises"""
        if len(img.shape) == 3:
            return cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        return img

    @staticmethod
    def a_hsv(img):
        """Convierte a espacio HSV"""
        return cv2.cvtColor(img, cv2.COLOR_RGB2HSV)

    @staticmethod
    def a_lab(img):
        """Convierte a espacio LAB"""
        return cv2.cvtColor(img, cv2.COLOR_RGB2LAB)

    @staticmethod
    def a_ycrcb(img):
        """Convierte a espacio YCrCb"""
        return cv2.cvtColor(img, cv2.COLOR_RGB2YCrCb)

    # =========================================================================
    # EXTRACCIÓN DE CANALES
    # =========================================================================

    @staticmethod
    def canal_rojo(img):
        """Extrae canal rojo"""
        canal = np.zeros_like(img)
        canal[:,:,0] = img[:,:,0]
        return canal

    @staticmethod
    def canal_verde(img):
        """Extrae canal verde"""
        canal = np.zeros_like(img)
        canal[:,:,1] = img[:,:,1]
        return canal

    @staticmethod
    def canal_azul(img):
        """Extrae canal azul"""
        canal = np.zeros_like(img)
        canal[:,:,2] = img[:,:,2]
        return canal

    @staticmethod
    def canal_rojo_gray(img):
        """Canal rojo como escala de grises"""
        return img[:,:,0]

    @staticmethod
    def canal_verde_gray(img):
        """Canal verde como escala de grises"""
        return img[:,:,1]

    @staticmethod
    def canal_azul_gray(img):
        """Canal azul como escala de grises"""
        return img[:,:,2]

    @staticmethod
    def canal_h(img):
        """Extrae canal Hue (tono) de HSV"""
        hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)
        return hsv[:,:,0]

    @staticmethod
    def canal_s(img):
        """Extrae canal Saturación de HSV"""
        hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)
        return hsv[:,:,1]

    @staticmethod
    def canal_v(img):
        """Extrae canal Value (brillo) de HSV"""
        hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)
        return hsv[:,:,2]

    @staticmethod
    def canal_l(img):
        """Extrae canal L (luminosidad) de LAB"""
        lab = cv2.cvtColor(img, cv2.COLOR_RGB2LAB)
        return lab[:,:,0]

    @staticmethod
    def canal_a(img):
        """Extrae canal A de LAB (verde-rojo)"""
        lab = cv2.cvtColor(img, cv2.COLOR_RGB2LAB)
        return lab[:,:,1]

    @staticmethod
    def canal_b_lab(img):
        """Extrae canal B de LAB (azul-amarillo)"""
        lab = cv2.cvtColor(img, cv2.COLOR_RGB2LAB)
        return lab[:,:,2]

    # =========================================================================
    # FILTROS DE SUAVIZADO (REDUCCIÓN DE RUIDO)
    # =========================================================================

    @staticmethod
    def filtro_media(img, k=5):
        """Filtro de media (promedio) - Suaviza la imagen"""
        k = k if k % 2 == 1 else k + 1
        return cv2.blur(img, (k, k))

    @staticmethod
    def filtro_gaussiano(img, k=5, sigma=0):
        """Filtro Gaussiano - Suavizado que preserva mejor los bordes"""
        k = k if k % 2 == 1 else k + 1
        return cv2.GaussianBlur(img, (k, k), sigma)

    @staticmethod
    def filtro_mediana(img, k=5):
        """Filtro de Mediana - Excelente para ruido sal y pimienta"""
        k = k if k % 2 == 1 else k + 1
        return cv2.medianBlur(img, k)

    @staticmethod
    def filtro_bilateral(img, d=9, sigma_color=75, sigma_space=75):
        """Filtro Bilateral - Suaviza preservando bordes nítidos"""
        return cv2.bilateralFilter(img, d, sigma_color, sigma_space)

    @staticmethod
    def filtro_nlm(img, h=10, template_size=7, search_size=21):
        """Non-Local Means - Reducción de ruido avanzada"""
        if len(img.shape) == 3:
            return cv2.fastNlMeansDenoisingColored(img, None, h, h, template_size, search_size)
        return cv2.fastNlMeansDenoising(img, None, h, template_size, search_size)

    # =========================================================================
    # MEJORA DE CONTRASTE
    # =========================================================================

    @staticmethod
    def ecualizar_histograma(img):
        """Ecualización de histograma global"""
        if len(img.shape) == 3:
            yuv = cv2.cvtColor(img, cv2.COLOR_RGB2YUV)
            yuv[:,:,0] = cv2.equalizeHist(yuv[:,:,0])
            return cv2.cvtColor(yuv, cv2.COLOR_YUV2RGB)
        return cv2.equalizeHist(img)

    @staticmethod
    def clahe(img, clip_limit=2.0, tile_size=8):
        """CLAHE - Ecualización adaptativa (mejor para billetes)"""
        clahe_obj = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=(tile_size, tile_size))
        if len(img.shape) == 3:
            lab = cv2.cvtColor(img, cv2.COLOR_RGB2LAB)
            lab[:,:,0] = clahe_obj.apply(lab[:,:,0])
            return cv2.cvtColor(lab, cv2.COLOR_LAB2RGB)
        return clahe_obj.apply(img)

    @staticmethod
    def correccion_gamma(img, gamma=1.0):
        """Corrección gamma para ajustar brillo"""
        inv_gamma = 1.0 / gamma
        tabla = np.array([((i / 255.0) ** inv_gamma) * 255 for i in range(256)]).astype(np.uint8)
        return cv2.LUT(img, tabla)

    @staticmethod
    def ajustar_brillo_contraste(img, brillo=0, contraste=1.0):
        """Ajuste manual de brillo y contraste"""
        resultado = contraste * img.astype(np.float32) + brillo
        return np.clip(resultado, 0, 255).astype(np.uint8)

    @staticmethod
    def stretching_contraste(img, p_low=2, p_high=98):
        """Estiramiento de contraste por percentiles"""
        if len(img.shape) == 3:
            resultado = np.zeros_like(img)
            for i in range(3):
                p2, p98 = np.percentile(img[:,:,i], (p_low, p_high))
                if p98 > p2:
                    resultado[:,:,i] = np.clip((img[:,:,i] - p2) * 255.0 / (p98 - p2), 0, 255)
            return resultado.astype(np.uint8)
        p2, p98 = np.percentile(img, (p_low, p_high))
        if p98 > p2:
            return np.clip((img - p2) * 255.0 / (p98 - p2), 0, 255).astype(np.uint8)
        return img

    # =========================================================================
    # REALCE Y NITIDEZ
    # =========================================================================

    @staticmethod
    def unsharp_masking(img, sigma=1.0, strength=1.5):
        """Máscara de enfoque - Aumenta nitidez"""
        blurred = cv2.GaussianBlur(img, (0, 0), sigma)
        sharpened = cv2.addWeighted(img, 1.0 + strength, blurred, -strength, 0)
        return np.clip(sharpened, 0, 255).astype(np.uint8)

    @staticmethod
    def high_boost(img, k=1.5):
        """Filtro High-Boost para realce"""
        blurred = cv2.GaussianBlur(img, (5, 5), 0)
        mask = img.astype(np.float32) - blurred.astype(np.float32)
        resultado = img.astype(np.float32) + k * mask
        return np.clip(resultado, 0, 255).astype(np.uint8)

    # =========================================================================
    # DETECCIÓN DE BORDES (CRUCIAL PARA DETECTAR FALSIFICACIONES)
    # =========================================================================

    @staticmethod
    def canny(img, t1=50, t2=150):
        """Detector de bordes Canny - El más preciso"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        return cv2.Canny(gray, t1, t2)

    @staticmethod
    def sobel(img, ksize=3):
        """Operador Sobel - Magnitud del gradiente"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=ksize)
        sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=ksize)
        magnitude = np.sqrt(sobelx**2 + sobely**2)
        return PDI.normalizar(magnitude)

    @staticmethod
    def sobel_x(img, ksize=3):
        """Sobel en dirección X (bordes verticales)"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=ksize)
        return PDI.normalizar(np.abs(sobelx))

    @staticmethod
    def sobel_y(img, ksize=3):
        """Sobel en dirección Y (bordes horizontales)"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=ksize)
        return PDI.normalizar(np.abs(sobely))

    @staticmethod
    def laplaciano(img, ksize=3):
        """Operador Laplaciano - Detecta cambios rápidos"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        lap = cv2.Laplacian(gray, cv2.CV_64F, ksize=ksize)
        return PDI.normalizar(np.abs(lap))

    @staticmethod
    def prewitt(img):
        """Operador Prewitt"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        kernelx = np.array([[1, 0, -1], [1, 0, -1], [1, 0, -1]], dtype=np.float32)
        kernely = np.array([[1, 1, 1], [0, 0, 0], [-1, -1, -1]], dtype=np.float32)
        prewittx = cv2.filter2D(gray, -1, kernelx)
        prewitty = cv2.filter2D(gray, -1, kernely)
        magnitude = np.sqrt(prewittx.astype(np.float32)**2 + prewitty.astype(np.float32)**2)
        return PDI.normalizar(magnitude)

    @staticmethod
    def scharr(img):
        """Operador Scharr - Más preciso que Sobel"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        scharrx = cv2.Scharr(gray, cv2.CV_64F, 1, 0)
        scharry = cv2.Scharr(gray, cv2.CV_64F, 0, 1)
        magnitude = np.sqrt(scharrx**2 + scharry**2)
        return PDI.normalizar(magnitude)

    @staticmethod
    def roberts(img):
        """Operador Roberts Cross"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        kernelx = np.array([[1, 0], [0, -1]], dtype=np.float32)
        kernely = np.array([[0, 1], [-1, 0]], dtype=np.float32)
        robertsx = cv2.filter2D(gray.astype(np.float32), -1, kernelx)
        robertsy = cv2.filter2D(gray.astype(np.float32), -1, kernely)
        magnitude = np.sqrt(robertsx**2 + robertsy**2)
        return PDI.normalizar(magnitude)

    @staticmethod
    def log(img, sigma=1.0):
        """Laplacian of Gaussian (LoG)"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        blurred = cv2.GaussianBlur(gray, (0, 0), sigma)
        log_result = cv2.Laplacian(blurred, cv2.CV_64F)
        return PDI.normalizar(np.abs(log_result))

    # =========================================================================
    # SEGMENTACIÓN Y BINARIZACIÓN
    # =========================================================================

    @staticmethod
    def umbral_simple(img, threshold=127):
        """Umbralización simple"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        _, binary = cv2.threshold(gray, threshold, 255, cv2.THRESH_BINARY)
        return binary

    @staticmethod
    def otsu(img):
        """Binarización Otsu - Umbral automático óptimo"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        _, binary = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
        return binary

    @staticmethod
    def adaptativo_gaussian(img, block_size=11, C=2):
        """Umbralización adaptativa Gaussiana"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        block_size = block_size if block_size % 2 == 1 else block_size + 1
        return cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                     cv2.THRESH_BINARY, block_size, C)

    @staticmethod
    def adaptativo_mean(img, block_size=11, C=2):
        """Umbralización adaptativa por media"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        block_size = block_size if block_size % 2 == 1 else block_size + 1
        return cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C,
                                     cv2.THRESH_BINARY, block_size, C)

    # =========================================================================
    # OPERACIONES MORFOLÓGICAS (IMPORTANTES PARA ANÁLISIS DE TEXTURA)
    # =========================================================================

    @staticmethod
    def erosion(img, kernel_size=3, iterations=1):
        """Erosión - Reduce regiones blancas"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        kernel = np.ones((kernel_size, kernel_size), np.uint8)
        return cv2.erode(gray, kernel, iterations=iterations)

    @staticmethod
    def dilatacion(img, kernel_size=3, iterations=1):
        """Dilatación - Expande regiones blancas"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        kernel = np.ones((kernel_size, kernel_size), np.uint8)
        return cv2.dilate(gray, kernel, iterations=iterations)

    @staticmethod
    def apertura(img, kernel_size=3):
        """Apertura - Elimina ruido pequeño"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        kernel = np.ones((kernel_size, kernel_size), np.uint8)
        return cv2.morphologyEx(gray, cv2.MORPH_OPEN, kernel)

    @staticmethod
    def cierre(img, kernel_size=3):
        """Cierre - Rellena huecos pequeños"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        kernel = np.ones((kernel_size, kernel_size), np.uint8)
        return cv2.morphologyEx(gray, cv2.MORPH_CLOSE, kernel)

    @staticmethod
    def gradiente_morfologico(img, kernel_size=3):
        """Gradiente morfológico - Contornos"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        kernel = np.ones((kernel_size, kernel_size), np.uint8)
        return cv2.morphologyEx(gray, cv2.MORPH_GRADIENT, kernel)

    @staticmethod
    def top_hat(img, kernel_size=9):
        """Top Hat - Resalta elementos brillantes pequeños"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        kernel = np.ones((kernel_size, kernel_size), np.uint8)
        return cv2.morphologyEx(gray, cv2.MORPH_TOPHAT, kernel)

    @staticmethod
    def black_hat(img, kernel_size=9):
        """Black Hat - Resalta elementos oscuros pequeños"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        kernel = np.ones((kernel_size, kernel_size), np.uint8)
        return cv2.morphologyEx(gray, cv2.MORPH_BLACKHAT, kernel)

    # =========================================================================
    # DETECCIÓN DE CARACTERÍSTICAS (PUNTOS CLAVE)
    # =========================================================================

    @staticmethod
    def detectar_contornos(img, modo='external'):
        """Detecta y dibuja contornos"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        _, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
        modo_cv = cv2.RETR_EXTERNAL if modo == 'external' else cv2.RETR_TREE
        contours, _ = cv2.findContours(binary, modo_cv, cv2.CHAIN_APPROX_SIMPLE)
        resultado = np.zeros((*gray.shape, 3), dtype=np.uint8)
        cv2.drawContours(resultado, contours, -1, (0, 255, 0), 2)
        return resultado

    @staticmethod
    def esquinas_harris(img, block_size=2, ksize=3, k=0.04):
        """Detector de esquinas Harris"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
            resultado = img.copy()
        else:
            gray = img
            resultado = cv2.cvtColor(gray, cv2.COLOR_GRAY2RGB)
        gray_float = np.float32(gray)
        dst = cv2.cornerHarris(gray_float, block_size, ksize, k)
        dst = cv2.dilate(dst, None)
        resultado[dst > 0.01 * dst.max()] = [255, 0, 0]
        return resultado

    @staticmethod
    def detector_orb(img, n_features=500):
        """Detector ORB - Puntos clave robustos"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
            resultado = img.copy()
        else:
            gray = img
            resultado = cv2.cvtColor(gray, cv2.COLOR_GRAY2RGB)
        orb = cv2.ORB_create(nfeatures=n_features)
        keypoints = orb.detect(gray, None)
        cv2.drawKeypoints(resultado, keypoints, resultado, color=(0, 255, 0),
                         flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)
        return resultado

    @staticmethod
    def hough_lineas(img, threshold=100):
        """Detecta líneas con transformada de Hough"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
            resultado = img.copy()
        else:
            gray = img
            resultado = cv2.cvtColor(gray, cv2.COLOR_GRAY2RGB)
        edges = cv2.Canny(gray, 50, 150)
        lines = cv2.HoughLinesP(edges, 1, np.pi/180, threshold, minLineLength=50, maxLineGap=10)
        if lines is not None:
            for line in lines:
                x1, y1, x2, y2 = line[0]
                cv2.line(resultado, (x1, y1), (x2, y2), (255, 0, 0), 2)
        return resultado

    # =========================================================================
    # TRANSFORMACIONES GEOMÉTRICAS
    # =========================================================================

    @staticmethod
    def invertir(img):
        """Invierte colores (negativo)"""
        return 255 - img

    @staticmethod
    def rotar(img, angulo):
        """Rota la imagen"""
        h, w = img.shape[:2]
        centro = (w // 2, h // 2)
        M = cv2.getRotationMatrix2D(centro, angulo, 1.0)
        return cv2.warpAffine(img, M, (w, h))

    @staticmethod
    def flip_horizontal(img):
        """Voltea horizontalmente"""
        return cv2.flip(img, 1)

    @staticmethod
    def flip_vertical(img):
        """Voltea verticalmente"""
        return cv2.flip(img, 0)

    @staticmethod
    def redimensionar(img, size=(224, 224)):
        """Redimensiona la imagen"""
        return cv2.resize(img, size, interpolation=cv2.INTER_AREA)

    # =========================================================================
    # ANÁLISIS ESTADÍSTICO DE TEXTURA
    # =========================================================================

    @staticmethod
    def extraer_caracteristicas(img):
        """Extrae características estadísticas de textura"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img

        # Estadísticas básicas
        media = np.mean(gray)
        desviacion = np.std(gray)
        varianza = np.var(gray)
        minimo = np.min(gray)
        maximo = np.max(gray)
        contraste = maximo - minimo

        # Entropía
        hist = cv2.calcHist([gray], [0], None, [256], [0, 256]).flatten()
        hist_norm = hist / (hist.sum() + 1e-10)
        entropia = -np.sum(hist_norm * np.log2(hist_norm + 1e-10))

        # Energía del gradiente
        sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
        sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
        energia = np.mean(np.sqrt(sobelx**2 + sobely**2))

        return {
            'Media': round(media, 2),
            'Desviación': round(desviacion, 2),
            'Varianza': round(varianza, 2),
            'Mínimo': int(minimo),
            'Máximo': int(maximo),
            'Contraste': int(contraste),
            'Entropía': round(entropia, 4),
            'Energía': round(energia, 2)
        }

    # =========================================================================
    # ANÁLISIS DE HISTOGRAMA
    # =========================================================================

    @staticmethod
    def calcular_histograma_rgb(img):
        """Calcula histograma para canales RGB"""
        histogramas = {}
        for i, color in enumerate(['R', 'G', 'B']):
            histogramas[color] = cv2.calcHist([img], [i], None, [256], [0, 256]).flatten()
        return histogramas

    @staticmethod
    def calcular_histograma_gray(img):
        """Calcula histograma en escala de grises"""
        if len(img.shape) == 3:
            gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        else:
            gray = img
        return cv2.calcHist([gray], [0], None, [256], [0, 256]).flatten()


# Crear diccionario de técnicas con nombres descriptivos
TECNICAS_PDI = {
    # Básicas
    "Original": lambda img: img,
    "Escala de Grises": lambda img: PDI.asegurar_rgb(PDI.a_gris(img)),

    # Canales de color
    "Canal Rojo": lambda img: PDI.canal_rojo(img),
    "Canal Verde": lambda img: PDI.canal_verde(img),
    "Canal Azul": lambda img: PDI.canal_azul(img),
    "Canal Hue (HSV)": lambda img: cv2.applyColorMap(PDI.canal_h(img), cv2.COLORMAP_HSV),
    "Canal Saturación (HSV)": lambda img: PDI.asegurar_rgb(PDI.canal_s(img)),
    "Canal Value (HSV)": lambda img: PDI.asegurar_rgb(PDI.canal_v(img)),
    "Canal L (LAB)": lambda img: PDI.asegurar_rgb(PDI.canal_l(img)),

    # Filtros de suavizado
    "Filtro Media k=5": lambda img: PDI.filtro_media(img, 5),
    "Filtro Gaussiano k=5": lambda img: PDI.filtro_gaussiano(img, 5),
    "Filtro Mediana k=5": lambda img: PDI.filtro_mediana(img, 5),
    "Filtro Bilateral": lambda img: PDI.filtro_bilateral(img),

    # Mejora de contraste
    "Ecualización Histograma": lambda img: PDI.ecualizar_histograma(img),
    "CLAHE": lambda img: PDI.clahe(img),
    "Gamma 0.5 (oscurecer)": lambda img: PDI.correccion_gamma(img, 0.5),
    "Gamma 2.0 (aclarar)": lambda img: PDI.correccion_gamma(img, 2.0),
    "Stretching Contraste": lambda img: PDI.stretching_contraste(img),

    # Realce
    "Unsharp Masking": lambda img: PDI.unsharp_masking(img),
    "High Boost": lambda img: PDI.high_boost(img),

    # Detección de bordes
    "Canny t1=50 t2=150": lambda img: PDI.asegurar_rgb(PDI.canny(img, 50, 150)),
    "Canny t1=100 t2=200": lambda img: PDI.asegurar_rgb(PDI.canny(img, 100, 200)),
    "Sobel (magnitud)": lambda img: PDI.asegurar_rgb(PDI.sobel(img)),
    "Sobel X (vertical)": lambda img: PDI.asegurar_rgb(PDI.sobel_x(img)),
    "Sobel Y (horizontal)": lambda img: PDI.asegurar_rgb(PDI.sobel_y(img)),
    "Laplaciano": lambda img: PDI.asegurar_rgb(PDI.laplaciano(img)),
    "Prewitt": lambda img: PDI.asegurar_rgb(PDI.prewitt(img)),
    "Scharr": lambda img: PDI.asegurar_rgb(PDI.scharr(img)),
    "Roberts": lambda img: PDI.asegurar_rgb(PDI.roberts(img)),
    "LoG (Laplacian of Gaussian)": lambda img: PDI.asegurar_rgb(PDI.log(img)),

    # Segmentación
    "Umbral Simple t=127": lambda img: PDI.asegurar_rgb(PDI.umbral_simple(img, 127)),
    "Binarización Otsu": lambda img: PDI.asegurar_rgb(PDI.otsu(img)),
    "Adaptativo Gaussiano": lambda img: PDI.asegurar_rgb(PDI.adaptativo_gaussian(img)),
    "Adaptativo Media": lambda img: PDI.asegurar_rgb(PDI.adaptativo_mean(img)),

    # Morfología
    "Erosión": lambda img: PDI.asegurar_rgb(PDI.erosion(img)),
    "Dilatación": lambda img: PDI.asegurar_rgb(PDI.dilatacion(img)),
    "Apertura": lambda img: PDI.asegurar_rgb(PDI.apertura(img)),
    "Cierre": lambda img: PDI.asegurar_rgb(PDI.cierre(img)),
    "Gradiente Morfológico": lambda img: PDI.asegurar_rgb(PDI.gradiente_morfologico(img)),
    "Top Hat": lambda img: PDI.asegurar_rgb(PDI.top_hat(img)),
    "Black Hat": lambda img: PDI.asegurar_rgb(PDI.black_hat(img)),

    # Detección de características
    "Contornos": lambda img: PDI.detectar_contornos(img),
    "Esquinas Harris": lambda img: PDI.esquinas_harris(img),
    "Detector ORB": lambda img: PDI.detector_orb(img),
    "Líneas Hough": lambda img: PDI.hough_lineas(img),

    # Transformaciones
    "Invertir (negativo)": lambda img: PDI.invertir(img),
    "Flip Horizontal": lambda img: PDI.flip_horizontal(img),
}

print(f"✅ Clase PDI inicializada con {len(TECNICAS_PDI)} técnicas disponibles")
print("\n📋 TÉCNICAS DISPONIBLES:")
for i, nombre in enumerate(TECNICAS_PDI.keys(), 1):
    print(f"   {i:2d}. {nombre}")
print("="*70)

# =============================================================================
# CELDA 5: FUNCIONES AUXILIARES
# =============================================================================

print("\n" + "="*70)
print("🛠️ INICIALIZANDO FUNCIONES AUXILIARES")
print("="*70)

def cargar_imagen(categoria, dataset_path=TRAIN_PATH, indice=None):
    """
    Carga una imagen de una categoría específica.
    Si indice es None, selecciona una aleatoria.
    """
    cat_path = os.path.join(dataset_path, categoria)
    if not os.path.exists(cat_path):
        print(f"⚠️ Categoría no encontrada: {categoria}")
        return None

    imagenes = [f for f in os.listdir(cat_path) if f.endswith(('.jpg', '.jpeg', '.png'))]
    if not imagenes:
        print(f"⚠️ No hay imágenes en: {categoria}")
        return None

    if indice is None:
        img_nombre = random.choice(imagenes)
    else:
        indice = min(indice, len(imagenes) - 1)
        img_nombre = imagenes[indice]

    img_path = os.path.join(cat_path, img_nombre)
    img = cv2.imread(img_path)

    if img is None:
        print(f"⚠️ Error cargando: {img_path}")
        return None

    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


def obtener_info_categoria(categoria):
    """Obtiene información de la categoría del billete"""
    # Extraer denominación y tipo de la categoría
    partes = categoria.lower().replace('_anverso', '').replace('_reverso', '').split('_')

    denom = partes[0] if partes else None
    tipo = 'nuevo' if 'nuevo' in categoria.lower() or 'new' in categoria.lower() else 'antiguo'
    lado = 'Anverso' if 'anverso' in categoria.lower() else 'Reverso' if 'reverso' in categoria.lower() else 'N/A'

    info = {
        'categoria': categoria,
        'denominacion': f"S/ {denom}",
        'tipo': tipo.capitalize(),
        'lado': lado
    }

    # Buscar información adicional
    if denom in INFO_BILLETES:
        if tipo in INFO_BILLETES[denom]:
            info.update(INFO_BILLETES[denom][tipo])

    return info


def mostrar_imagen(img, titulo="Imagen", figsize=(10, 6)):
    """Muestra una imagen con matplotlib"""
    plt.figure(figsize=figsize)
    if len(img.shape) == 2:
        plt.imshow(img, cmap='gray')
    else:
        plt.imshow(img)
    plt.title(titulo, fontsize=14, fontweight='bold')
    plt.axis('off')
    plt.tight_layout()
    plt.show()


def crear_grid_imagenes(imagenes, titulos, filas, columnas, figsize=(16, 12), titulo_principal=None):
    """Crea una cuadrícula de imágenes"""
    fig, axes = plt.subplots(filas, columnas, figsize=figsize)
    if titulo_principal:
        fig.suptitle(titulo_principal, fontsize=16, fontweight='bold', y=1.02)

    axes = axes.flatten() if filas * columnas > 1 else [axes]

    for i, (img, titulo) in enumerate(zip(imagenes, titulos)):
        if i < len(axes):
            if img is not None:
                if len(img.shape) == 2:
                    axes[i].imshow(img, cmap='gray')
                else:
                    axes[i].imshow(img)
            axes[i].set_title(titulo, fontsize=10, fontweight='bold')
            axes[i].axis('off')

    # Ocultar ejes vacíos
    for i in range(len(imagenes), len(axes)):
        axes[i].axis('off')

    plt.tight_layout()
    plt.show()
    return fig


print("✅ Funciones auxiliares cargadas")
print("="*70)

# =============================================================================
# CELDA 6: VISUALIZACIÓN DEL DATASET - MUESTRAS POR CATEGORÍA
# =============================================================================

print("\n" + "="*70)
print(" OBJETIVO A: ANÁLISIS DE CARACTERÍSTICAS VISUALES")
print("   Mostrando muestras de cada categoría del dataset")
print("="*70)

# Mostrar una imagen de cada categoría
fig, axes = plt.subplots(4, 4, figsize=(16, 16))
fig.suptitle('🏦 MUESTRAS DEL DATASET DE BILLETES PERUANOS\n(Una imagen por categoría)',
             fontsize=16, fontweight='bold')

for i, categoria in enumerate(CATEGORIAS[:16]):
    row, col = i // 4, i % 4
    img = cargar_imagen(categoria)
    if img is not None:
        img_resized = PDI.redimensionar(img, (300, 150))
        axes[row, col].imshow(img_resized)
        # Formatear título
        titulo = categoria.replace('_', ' ').title()
        if len(titulo) > 20:
            titulo = titulo[:18] + '...'
        axes[row, col].set_title(titulo, fontsize=9, fontweight='bold')
    axes[row, col].axis('off')

plt.tight_layout()
plt.savefig('/content/01_muestras_dataset.png', dpi=150, bbox_inches='tight')
plt.show()
print(" Figura guardada: 01_muestras_dataset.png")

# =============================================================================
# CELDA 7: DEMOSTRACIÓN COMPLETA DE PDI EN UN BILLETE
# =============================================================================

print("\n" + "="*70)
print("🔬 DEMOSTRACIÓN COMPLETA DE TÉCNICAS PDI")
print("   Aplicando todas las técnicas a un billete de muestra")
print("="*70)

# Seleccionar un billete de muestra
categoria_demo = CATEGORIAS[0]
img_demo = cargar_imagen(categoria_demo)

if img_demo is not None:
    img_demo = PDI.redimensionar(img_demo, (400, 200))

    print(f"\n Billete seleccionado: {categoria_demo}")
    print(f"   Dimensiones: {img_demo.shape}")

    # Mostrar características de textura
    caracteristicas = PDI.extraer_caracteristicas(img_demo)
    print(f"\n Características de textura:")
    for k, v in caracteristicas.items():
        print(f"   • {k}: {v}")

# =============================================================================
# SECCIÓN 7.1: FILTROS DE SUAVIZADO
# =============================================================================

print("\n" + "-"*50)
print(" SECCIÓN 1: FILTROS DE SUAVIZADO")
print("-"*50)

fig, axes = plt.subplots(2, 4, figsize=(18, 9))
fig.suptitle(' FILTROS DE SUAVIZADO\nReducción de ruido preservando características importantes',
             fontsize=14, fontweight='bold')

filtros_suavizado = [
    ("Original", img_demo),
    ("Filtro Media 5x5", PDI.filtro_media(img_demo, 5)),
    ("Filtro Media 9x9", PDI.filtro_media(img_demo, 9)),
    ("Filtro Gaussiano 5x5", PDI.filtro_gaussiano(img_demo, 5)),
    ("Filtro Gaussiano 9x9", PDI.filtro_gaussiano(img_demo, 9)),
    ("Filtro Mediana 5x5", PDI.filtro_mediana(img_demo, 5)),
    ("Filtro Bilateral", PDI.filtro_bilateral(img_demo)),
    ("Non-Local Means", PDI.filtro_nlm(img_demo)),
]

for i, (titulo, img) in enumerate(filtros_suavizado):
    row, col = i // 4, i % 4
    axes[row, col].imshow(img)
    axes[row, col].set_title(titulo, fontsize=11, fontweight='bold')
    axes[row, col].axis('off')

plt.tight_layout()
plt.savefig('/content/02_pdi_filtros_suavizado.png', dpi=150, bbox_inches='tight')
plt.show()
print(" Figura guardada: 02_pdi_filtros_suavizado.png")

# =============================================================================
# SECCIÓN 7.2: MEJORA DE CONTRASTE
# =============================================================================

print("\n" + "-"*50)
print(" SECCIÓN 2: MEJORA DE CONTRASTE")
print("-"*50)

fig, axes = plt.subplots(2, 4, figsize=(18, 9))
fig.suptitle(' MEJORA DE CONTRASTE\nRealce de detalles para mejor análisis',
             fontsize=14, fontweight='bold')

mejora_contraste = [
    ("Original", img_demo),
    ("Ecualización Histograma", PDI.ecualizar_histograma(img_demo)),
    ("CLAHE (Adaptativo)", PDI.clahe(img_demo)),
    ("CLAHE clip=4.0", PDI.clahe(img_demo, clip_limit=4.0)),
    ("Gamma 0.5 (oscuro)", PDI.correccion_gamma(img_demo, 0.5)),
    ("Gamma 1.5 (claro)", PDI.correccion_gamma(img_demo, 1.5)),
    ("Gamma 2.0 (más claro)", PDI.correccion_gamma(img_demo, 2.0)),
    ("Stretching Contraste", PDI.stretching_contraste(img_demo)),
]

for i, (titulo, img) in enumerate(mejora_contraste):
    row, col = i // 4, i % 4
    axes[row, col].imshow(img)
    axes[row, col].set_title(titulo, fontsize=11, fontweight='bold')
    axes[row, col].axis('off')

plt.tight_layout()
plt.savefig('/content/03_pdi_mejora_contraste.png', dpi=150, bbox_inches='tight')
plt.show()
print(" Figura guardada: 03_pdi_mejora_contraste.png")

# =============================================================================
# SECCIÓN 7.3: DETECCIÓN DE BORDES
# =============================================================================

print("\n" + "-"*50)
print(" SECCIÓN 3: DETECCIÓN DE BORDES (Crucial para falsificaciones)")
print("-"*50)

fig, axes = plt.subplots(3, 4, figsize=(18, 13))
fig.suptitle(' DETECCIÓN DE BORDES\nIdentificación de contornos y patrones de seguridad',
             fontsize=14, fontweight='bold')

deteccion_bordes = [
    ("Original", img_demo),
    ("Escala de Grises", PDI.asegurar_rgb(PDI.a_gris(img_demo))),
    ("Canny t1=50 t2=150", PDI.asegurar_rgb(PDI.canny(img_demo, 50, 150))),
    ("Canny t1=100 t2=200", PDI.asegurar_rgb(PDI.canny(img_demo, 100, 200))),
    ("Sobel (magnitud)", PDI.asegurar_rgb(PDI.sobel(img_demo))),
    ("Sobel X (vertical)", PDI.asegurar_rgb(PDI.sobel_x(img_demo))),
    ("Sobel Y (horizontal)", PDI.asegurar_rgb(PDI.sobel_y(img_demo))),
    ("Laplaciano", PDI.asegurar_rgb(PDI.laplaciano(img_demo))),
    ("Prewitt", PDI.asegurar_rgb(PDI.prewitt(img_demo))),
    ("Scharr", PDI.asegurar_rgb(PDI.scharr(img_demo))),
    ("Roberts", PDI.asegurar_rgb(PDI.roberts(img_demo))),
    ("LoG (Laplacian of Gaussian)", PDI.asegurar_rgb(PDI.log(img_demo))),
]

for i, (titulo, img) in enumerate(deteccion_bordes):
    row, col = i // 4, i % 4
    if len(img.shape) == 2:
        axes[row, col].imshow(img, cmap='gray')
    else:
        axes[row, col].imshow(img)
    axes[row, col].set_title(titulo, fontsize=10, fontweight='bold')
    axes[row, col].axis('off')

plt.tight_layout()
plt.savefig('/content/04_pdi_deteccion_bordes.png', dpi=150, bbox_inches='tight')
plt.show()
print(" Figura guardada: 04_pdi_deteccion_bordes.png")

# =============================================================================
# SECCIÓN 7.4: SEGMENTACIÓN Y BINARIZACIÓN
# =============================================================================

print("\n" + "-"*50)
print(" SECCIÓN 4: SEGMENTACIÓN Y BINARIZACIÓN")
print("-"*50)

fig, axes = plt.subplots(2, 4, figsize=(18, 9))
fig.suptitle(' SEGMENTACIÓN Y BINARIZACIÓN\nSeparación de elementos del fondo',
             fontsize=14, fontweight='bold')

segmentacion = [
    ("Original", img_demo),
    ("Escala de Grises", PDI.asegurar_rgb(PDI.a_gris(img_demo))),
    ("Umbral Simple t=127", PDI.asegurar_rgb(PDI.umbral_simple(img_demo, 127))),
    ("Umbral Simple t=100", PDI.asegurar_rgb(PDI.umbral_simple(img_demo, 100))),
    ("Binarización Otsu", PDI.asegurar_rgb(PDI.otsu(img_demo))),
    ("Adaptativo Gaussiano", PDI.asegurar_rgb(PDI.adaptativo_gaussian(img_demo))),
    ("Adaptativo Media", PDI.asegurar_rgb(PDI.adaptativo_mean(img_demo))),
    ("Adaptativo bs=21", PDI.asegurar_rgb(PDI.adaptativo_gaussian(img_demo, block_size=21))),
]

for i, (titulo, img) in enumerate(segmentacion):
    row, col = i // 4, i % 4
    axes[row, col].imshow(img, cmap='gray' if len(img.shape)==2 else None)
    axes[row, col].set_title(titulo, fontsize=10, fontweight='bold')
    axes[row, col].axis('off')

plt.tight_layout()
plt.savefig('/content/05_pdi_segmentacion.png', dpi=150, bbox_inches='tight')
plt.show()
print(" Figura guardada: 05_pdi_segmentacion.png")

# =============================================================================
# SECCIÓN 7.5: OPERACIONES MORFOLÓGICAS
# =============================================================================

print("\n" + "-"*50)
print(" SECCIÓN 5: OPERACIONES MORFOLÓGICAS")
print("-"*50)

fig, axes = plt.subplots(2, 4, figsize=(18, 9))
fig.suptitle(' OPERACIONES MORFOLÓGICAS\nAnálisis de formas y texturas',
             fontsize=14, fontweight='bold')

morfologia = [
    ("Original (Grises)", PDI.asegurar_rgb(PDI.a_gris(img_demo))),
    ("Erosión", PDI.asegurar_rgb(PDI.erosion(img_demo))),
    ("Dilatación", PDI.asegurar_rgb(PDI.dilatacion(img_demo))),
    ("Apertura", PDI.asegurar_rgb(PDI.apertura(img_demo))),
    ("Cierre", PDI.asegurar_rgb(PDI.cierre(img_demo))),
    ("Gradiente Morfológico", PDI.asegurar_rgb(PDI.gradiente_morfologico(img_demo))),
    ("Top Hat", PDI.asegurar_rgb(PDI.top_hat(img_demo))),
    ("Black Hat", PDI.asegurar_rgb(PDI.black_hat(img_demo))),
]

for i, (titulo, img) in enumerate(morfologia):
    row, col = i // 4, i % 4
    axes[row, col].imshow(img, cmap='gray' if len(img.shape)==2 else None)
    axes[row, col].set_title(titulo, fontsize=10, fontweight='bold')
    axes[row, col].axis('off')

plt.tight_layout()
plt.savefig('/content/06_pdi_morfologia.png', dpi=150, bbox_inches='tight')
plt.show()
print(" Figura guardada: 06_pdi_morfologia.png")

# =============================================================================
# SECCIÓN 7.6: ANÁLISIS DE COLOR
# =============================================================================

print("\n" + "-"*50)
print(" SECCIÓN 6: ANÁLISIS DE COLOR")
print("-"*50)

fig, axes = plt.subplots(3, 4, figsize=(18, 13))
fig.suptitle(' ANÁLISIS DE COLOR\nDescomposición en diferentes espacios de color',
             fontsize=14, fontweight='bold')

analisis_color = [
    ("Original RGB", img_demo),
    ("Canal Rojo", PDI.canal_rojo(img_demo)),
    ("Canal Verde", PDI.canal_verde(img_demo)),
    ("Canal Azul", PDI.canal_azul(img_demo)),
    ("Canal R (gris)", PDI.asegurar_rgb(PDI.canal_rojo_gray(img_demo))),
    ("Canal G (gris)", PDI.asegurar_rgb(PDI.canal_verde_gray(img_demo))),
    ("Canal B (gris)", PDI.asegurar_rgb(PDI.canal_azul_gray(img_demo))),
    ("HSV Completo", PDI.a_hsv(img_demo)),
    ("Canal Hue", cv2.applyColorMap(PDI.canal_h(img_demo), cv2.COLORMAP_HSV)),
    ("Canal Saturación", PDI.asegurar_rgb(PDI.canal_s(img_demo))),
    ("Canal Value", PDI.asegurar_rgb(PDI.canal_v(img_demo))),
    ("Canal L (LAB)", PDI.asegurar_rgb(PDI.canal_l(img_demo))),
]

for i, (titulo, img) in enumerate(analisis_color):
    row, col = i // 4, i % 4
    axes[row, col].imshow(img)
    axes[row, col].set_title(titulo, fontsize=10, fontweight='bold')
    axes[row, col].axis('off')

plt.tight_layout()
plt.savefig('/content/07_pdi_analisis_color.png', dpi=150, bbox_inches='tight')
plt.show()
print(" Figura guardada: 07_pdi_analisis_color.png")

# =============================================================================
# SECCIÓN 7.7: DETECCIÓN DE CARACTERÍSTICAS
# =============================================================================

print("\n" + "-"*50)
print(" SECCIÓN 7: DETECCIÓN DE CARACTERÍSTICAS")
print("-"*50)

fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle(' DETECCIÓN DE CARACTERÍSTICAS\nPuntos clave, contornos y líneas',
             fontsize=14, fontweight='bold')

caracteristicas = [
    ("Original", img_demo),
    ("Contornos Detectados", PDI.detectar_contornos(img_demo)),
    ("Esquinas Harris", PDI.esquinas_harris(img_demo)),
    ("Detector ORB (500 pts)", PDI.detector_orb(img_demo, 500)),
    ("Líneas Hough", PDI.hough_lineas(img_demo)),
    ("Invertir (Negativo)", PDI.invertir(img_demo)),
]

for i, (titulo, img) in enumerate(caracteristicas):
    row, col = i // 3, i % 3
    axes[row, col].imshow(img)
    axes[row, col].set_title(titulo, fontsize=11, fontweight='bold')
    axes[row, col].axis('off')

plt.tight_layout()
plt.savefig('/content/08_pdi_caracteristicas.png', dpi=150, bbox_inches='tight')
plt.show()
print(" Figura guardada: 08_pdi_caracteristicas.png")

# =============================================================================
# CELDA 8: COMPARATIVA PDI POR DENOMINACIÓN
# =============================================================================

print("\n" + "="*70)
print(" COMPARATIVA PDI POR DENOMINACIÓN")
print("   Aplicando técnicas clave a cada tipo de billete")
print("="*70)

fig, axes = plt.subplots(4, 7, figsize=(22, 14))
fig.suptitle(' COMPARATIVA DE TÉCNICAS PDI POR DENOMINACIÓN\n'
             'Análisis para detección de falsificaciones', fontsize=16, fontweight='bold')

# Títulos de columnas
columnas = ['Original', 'CLAHE', 'Canny', 'Sobel', 'Otsu', 'Gradiente Morf.', 'Harris']

# Añadir títulos de columnas
for j, titulo in enumerate(columnas):
    axes[0, j].set_title(titulo, fontsize=11, fontweight='bold', pad=10)

denominaciones = ['10', '20', '50', '100']

for i, denom in enumerate(denominaciones):
    # Buscar categoría que contenga esta denominación
    cat = None
    for c in CATEGORIAS:
        if c.startswith(denom):
            cat = c
            break

    if cat:
        img = cargar_imagen(cat)
        if img is not None:
            img = PDI.redimensionar(img, (350, 175))

            # Aplicar técnicas
            resultados = [
                img,                                              # Original
                PDI.clahe(img),                                   # CLAHE
                PDI.asegurar_rgb(PDI.canny(img)),                # Canny
                PDI.asegurar_rgb(PDI.sobel(img)),                # Sobel
                PDI.asegurar_rgb(PDI.otsu(img)),                 # Otsu
                PDI.asegurar_rgb(PDI.gradiente_morfologico(img)), # Gradiente
                PDI.esquinas_harris(img),                         # Harris
            ]

            # Mostrar
            axes[i, 0].set_ylabel(f'S/ {denom}', fontsize=14, fontweight='bold', rotation=0, ha='right', va='center')
            for j, resultado in enumerate(resultados):
                axes[i, j].imshow(resultado)
                axes[i, j].axis('off')

plt.tight_layout()
plt.savefig('/content/09_comparativa_denominaciones.png', dpi=150, bbox_inches='tight')
plt.show()
print(" Figura guardada: 09_comparativa_denominaciones.png")

# =============================================================================
# CELDA 9: ANÁLISIS DE HISTOGRAMAS POR DENOMINACIÓN
# =============================================================================

print("\n" + "="*70)
print(" ANÁLISIS DE HISTOGRAMAS RGB POR DENOMINACIÓN")
print("="*70)

fig, axes = plt.subplots(4, 2, figsize=(16, 16))
fig.suptitle(' HISTOGRAMAS RGB POR DENOMINACIÓN\n'
             'Distribución de colores característicos de cada billete', fontsize=14, fontweight='bold')

for i, denom in enumerate(denominaciones):
    cat = None
    for c in CATEGORIAS:
        if c.startswith(denom):
            cat = c
            break

    if cat:
        img = cargar_imagen(cat)
        if img is not None:
            img = PDI.redimensionar(img, (400, 200))

            # Mostrar imagen
            axes[i, 0].imshow(img)
            axes[i, 0].set_title(f'S/ {denom} - {cat}', fontsize=11, fontweight='bold')
            axes[i, 0].axis('off')

            # Calcular y mostrar histogramas
            histogramas = PDI.calcular_histograma_rgb(img)
            axes[i, 1].plot(histogramas['R'], color='red', alpha=0.7, label='Rojo', linewidth=2)
            axes[i, 1].plot(histogramas['G'], color='green', alpha=0.7, label='Verde', linewidth=2)
            axes[i, 1].plot(histogramas['B'], color='blue', alpha=0.7, label='Azul', linewidth=2)
            axes[i, 1].set_title(f'Histograma RGB - S/ {denom}', fontsize=11, fontweight='bold')
            axes[i, 1].set_xlabel('Intensidad (0-255)')
            axes[i, 1].set_ylabel('Frecuencia')
            axes[i, 1].legend()
            axes[i, 1].set_xlim([0, 255])
            axes[i, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('/content/10_histogramas_rgb.png', dpi=150, bbox_inches='tight')
plt.show()
print(" Figura guardada: 10_histogramas_rgb.png")

print("\n" + "="*70)
print(" ANÁLISIS PDI COMPLETADO")
print("   Se han generado 10 figuras con análisis detallado")
print("="*70)

# =============================================================================
# CELDA 10: MODELOS DE DEEP LEARNING
# =============================================================================

print("\n" + "="*70)
print(" OBJETIVO B y C: DEFINICIÓN DE ARQUITECTURAS DE DEEP LEARNING")
print("="*70)

IMG_SIZE = 224
BATCH_SIZE = 32

def crear_modelo_cnn(num_clases, nombre='CNN_Clasificador'):
    """
    CNN para clasificación de billetes.
    Arquitectura optimizada para extracción de características visuales.
    """
    model = Sequential([
        # Bloque 1
        Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(IMG_SIZE, IMG_SIZE, 3)),
        BatchNormalization(),
        Conv2D(32, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Dropout(0.25),

        # Bloque 2
        Conv2D(64, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        Conv2D(64, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Dropout(0.25),

        # Bloque 3
        Conv2D(128, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        Conv2D(128, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Dropout(0.25),

        # Bloque 4
        Conv2D(256, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        GlobalAveragePooling2D(),

        # Clasificador
        Dense(256, activation='relu'),
        Dropout(0.5),
        Dense(128, activation='relu'),
        Dropout(0.3),
        Dense(num_clases, activation='softmax')
    ], name=nombre)

    model.compile(
        optimizer=Adam(learning_rate=0.0001),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model


def crear_modelo_cnn_lstm(num_clases, nombre='CNN_LSTM_Hibrido'):
    """
    Modelo híbrido CNN-LSTM.
    CNN extrae características espaciales, LSTM procesa secuencialmente.
    """
    model = Sequential([
        # Extractor CNN
        Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(IMG_SIZE, IMG_SIZE, 3)),
        BatchNormalization(),
        MaxPooling2D((2, 2)),

        Conv2D(64, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),

        Conv2D(128, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),

        # Reshape para LSTM: (batch, timesteps, features)
        Reshape((28 * 28, 128)),

        # LSTM bidireccional
        Bidirectional(LSTM(64, return_sequences=True)),
        Dropout(0.3),
        Bidirectional(LSTM(32, return_sequences=False)),
        Dropout(0.3),

        # Clasificador
        Dense(128, activation='relu'),
        Dropout(0.5),
        Dense(num_clases, activation='softmax')
    ], name=nombre)

    model.compile(
        optimizer=Adam(learning_rate=0.0001),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model


def crear_modelo_cnn_gru(num_clases, nombre='CNN_GRU_Hibrido'):
    """
    Modelo híbrido CNN-GRU.
    Similar a CNN-LSTM pero con GRU (más eficiente).
    """
    model = Sequential([
        # Extractor CNN
        Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(IMG_SIZE, IMG_SIZE, 3)),
        BatchNormalization(),
        MaxPooling2D((2, 2)),

        Conv2D(64, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),

        Conv2D(128, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),

        # Reshape para GRU
        Reshape((28 * 28, 128)),

        # GRU bidireccional
        Bidirectional(GRU(64, return_sequences=True)),
        Dropout(0.3),
        Bidirectional(GRU(32, return_sequences=False)),
        Dropout(0.3),

        # Clasificador
        Dense(128, activation='relu'),
        Dropout(0.5),
        Dense(num_clases, activation='softmax')
    ], name=nombre)

    model.compile(
        optimizer=Adam(learning_rate=0.0001),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model


# Mostrar arquitecturas
print("\n ARQUITECTURAS DISPONIBLES:")
print("\n1️ CNN Clasificador:")
modelo_cnn_temp = crear_modelo_cnn(NUM_CLASES)
modelo_cnn_temp.summary()

print("\n2️ CNN-LSTM Híbrido:")
modelo_lstm_temp = crear_modelo_cnn_lstm(NUM_CLASES)
modelo_lstm_temp.summary()

print("\n3️ CNN-GRU Híbrido:")
modelo_gru_temp = crear_modelo_cnn_gru(NUM_CLASES)
modelo_gru_temp.summary()

# Limpiar memoria
del modelo_cnn_temp, modelo_lstm_temp, modelo_gru_temp
import gc
gc.collect()

print("\n Arquitecturas definidas correctamente")
print("="*70)

# =============================================================================
# CELDA 11: PREPARACIÓN DE DATOS PARA ENTRENAMIENTO
# =============================================================================

print("\n" + "="*70)
print(" PREPARACIÓN DE DATOS PARA ENTRENAMIENTO")
print("="*70)

# Generadores de datos con aumento
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=10,
    width_shift_range=0.1,
    height_shift_range=0.1,
    shear_range=0.1,
    zoom_range=0.1,
    brightness_range=[0.8, 1.2],
    horizontal_flip=False,  # No voltear billetes
    fill_mode='nearest'
)

val_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

print(" Cargando generadores de datos...")

train_generator = train_datagen.flow_from_directory(
    TRAIN_PATH,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=True
)

val_generator = val_datagen.flow_from_directory(
    VAL_PATH,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

test_generator = test_datagen.flow_from_directory(
    TEST_PATH,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

# Guardar mapeo de clases
class_indices = train_generator.class_indices
idx_to_class = {v: k for k, v in class_indices.items()}

print(f"\n Generadores creados:")
print(f"   Training:   {train_generator.samples} muestras")
print(f"   Validation: {val_generator.samples} muestras")
print(f"   Testing:    {test_generator.samples} muestras")
print(f"   Clases:     {len(class_indices)}")
print(f"   Batch size: {BATCH_SIZE}")

# Mostrar mapeo de clases
print(f"\n Mapeo de clases:")
for clase, idx in sorted(class_indices.items(), key=lambda x: x[1]):
    print(f"   {idx:2d}: {clase}")

print("="*70)

# =============================================================================
# CELDA 12: ENTRENAMIENTO DE MODELOS
# =============================================================================

print("\n" + "="*70)
print(" ENTRENAMIENTO DE MODELOS")
print("="*70)

# Parámetros de entrenamiento
EPOCHS = 20  # Ajustar según necesidad
PATIENCE_ES = 7
PATIENCE_LR = 4

# Callbacks
callbacks = [
    EarlyStopping(
        monitor='val_loss',
        patience=PATIENCE_ES,
        restore_best_weights=True,
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=PATIENCE_LR,
        verbose=1,
        min_lr=1e-7
    )
]

# Diccionario para almacenar resultados
resultados = {}

# -----------------------------------------------------------------------------
# ENTRENAMIENTO CNN
# -----------------------------------------------------------------------------
print("\n" + "-"*50)
print(" ENTRENANDO MODELO CNN...")
print("-"*50)

modelo_cnn = crear_modelo_cnn(NUM_CLASES)

historia_cnn = modelo_cnn.fit(
    train_generator,
    validation_data=val_generator,
    epochs=EPOCHS,
    callbacks=callbacks,
    verbose=1
)

resultados['CNN'] = {
    'modelo': modelo_cnn,
    'historia': historia_cnn.history,
    'train_acc': historia_cnn.history['accuracy'][-1],
    'val_acc': historia_cnn.history['val_accuracy'][-1]
}

print(f"\n CNN - Train Acc: {resultados['CNN']['train_acc']:.4f}, Val Acc: {resultados['CNN']['val_acc']:.4f}")

# -----------------------------------------------------------------------------
# ENTRENAMIENTO CNN-LSTM
# -----------------------------------------------------------------------------
print("\n" + "-"*50)
print("🏋️ ENTRENANDO MODELO CNN-LSTM...")
print("-"*50)

modelo_lstm = crear_modelo_cnn_lstm(NUM_CLASES)

historia_lstm = modelo_lstm.fit(
    train_generator,
    validation_data=val_generator,
    epochs=EPOCHS,
    callbacks=callbacks,
    verbose=1
)

resultados['CNN-LSTM'] = {
    'modelo': modelo_lstm,
    'historia': historia_lstm.history,
    'train_acc': historia_lstm.history['accuracy'][-1],
    'val_acc': historia_lstm.history['val_accuracy'][-1]
}

print(f"\n CNN-LSTM - Train Acc: {resultados['CNN-LSTM']['train_acc']:.4f}, Val Acc: {resultados['CNN-LSTM']['val_acc']:.4f}")

# -----------------------------------------------------------------------------
# ENTRENAMIENTO CNN-GRU
# -----------------------------------------------------------------------------
print("\n" + "-"*50)
print(" ENTRENANDO MODELO CNN-GRU...")
print("-"*50)

modelo_gru = crear_modelo_cnn_gru(NUM_CLASES)

historia_gru = modelo_gru.fit(
    train_generator,
    validation_data=val_generator,
    epochs=EPOCHS,
    callbacks=callbacks,
    verbose=1
)

resultados['CNN-GRU'] = {
    'modelo': modelo_gru,
    'historia': historia_gru.history,
    'train_acc': historia_gru.history['accuracy'][-1],
    'val_acc': historia_gru.history['val_accuracy'][-1]
}

print(f"\n CNN-GRU - Train Acc: {resultados['CNN-GRU']['train_acc']:.4f}, Val Acc: {resultados['CNN-GRU']['val_acc']:.4f}")

print("\n" + "="*70)
print(" ENTRENAMIENTO COMPLETADO")
print("="*70)

# =============================================================================
# CELDA 13: VISUALIZACIÓN DE CURVAS DE ENTRENAMIENTO
# =============================================================================

print("\n" + "="*70)
print(" CURVAS DE ENTRENAMIENTO")
print("="*70)

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle(' CURVAS DE ENTRENAMIENTO - COMPARATIVA DE MODELOS', fontsize=16, fontweight='bold')

colores = {'CNN': 'blue', 'CNN-LSTM': 'green', 'CNN-GRU': 'red'}

for i, (nombre, data) in enumerate(resultados.items()):
    historia = data['historia']
    color = colores[nombre]

    # Accuracy
    axes[0, i].plot(historia['accuracy'], f'{color[0]}-', label='Train', linewidth=2)
    axes[0, i].plot(historia['val_accuracy'], f'{color[0]}--', label='Validation', linewidth=2)
    axes[0, i].set_title(f'{nombre} - Accuracy', fontsize=12, fontweight='bold')
    axes[0, i].set_xlabel('Época')
    axes[0, i].set_ylabel('Accuracy')
    axes[0, i].legend()
    axes[0, i].grid(True, alpha=0.3)
    axes[0, i].set_ylim([0, 1])

    # Loss
    axes[1, i].plot(historia['loss'], f'{color[0]}-', label='Train', linewidth=2)
    axes[1, i].plot(historia['val_loss'], f'{color[0]}--', label='Validation', linewidth=2)
    axes[1, i].set_title(f'{nombre} - Loss', fontsize=12, fontweight='bold')
    axes[1, i].set_xlabel('Época')
    axes[1, i].set_ylabel('Loss')
    axes[1, i].legend()
    axes[1, i].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('/content/11_curvas_entrenamiento.png', dpi=150, bbox_inches='tight')
plt.show()
print(" Figura guardada: 11_curvas_entrenamiento.png")

# =============================================================================
# CELDA 14: EVALUACIÓN EN CONJUNTO DE TEST
# =============================================================================

print("\n" + "="*70)
print(" OBJETIVO D: EVALUACIÓN DE MODELOS EN CONJUNTO DE TEST")
print("="*70)

# Evaluar cada modelo
evaluaciones = {}

for nombre, data in resultados.items():
    modelo = data['modelo']

    # Resetear generador
    test_generator.reset()

    # Evaluar
    loss, accuracy = modelo.evaluate(test_generator, verbose=0)

    # Predicciones
    test_generator.reset()
    y_pred_proba = modelo.predict(test_generator, verbose=0)
    y_pred = np.argmax(y_pred_proba, axis=1)
    y_true = test_generator.classes

    # Métricas
    from sklearn.metrics import precision_score, recall_score, f1_score

    evaluaciones[nombre] = {
        'accuracy': accuracy,
        'loss': loss,
        'precision': precision_score(y_true, y_pred, average='weighted', zero_division=0),
        'recall': recall_score(y_true, y_pred, average='weighted', zero_division=0),
        'f1': f1_score(y_true, y_pred, average='weighted', zero_division=0),
        'y_true': y_true,
        'y_pred': y_pred,
        'y_pred_proba': y_pred_proba
    }

    print(f"\n📊 {nombre}:")
    print(f"   Accuracy:  {accuracy:.4f} ({accuracy*100:.2f}%)")
    print(f"   Loss:      {loss:.4f}")
    print(f"   Precision: {evaluaciones[nombre]['precision']:.4f}")
    print(f"   Recall:    {evaluaciones[nombre]['recall']:.4f}")
    print(f"   F1-Score:  {evaluaciones[nombre]['f1']:.4f}")

# Tabla comparativa
print("\n" + "-"*50)
print(" TABLA COMPARATIVA DE MODELOS")
print("-"*50)
print(f"{'Modelo':<12} {'Accuracy':>10} {'Precision':>10} {'Recall':>10} {'F1-Score':>10}")
print("-"*52)
for nombre, metricas in evaluaciones.items():
    print(f"{nombre:<12} {metricas['accuracy']:>10.4f} {metricas['precision']:>10.4f} {metricas['recall']:>10.4f} {metricas['f1']:>10.4f}")

# Mejor modelo
mejor_modelo = max(evaluaciones.keys(), key=lambda x: evaluaciones[x]['accuracy'])
print(f"\n MEJOR MODELO: {mejor_modelo} con {evaluaciones[mejor_modelo]['accuracy']*100:.2f}% de accuracy")

print("="*70)

# =============================================================================
# CELDA 15: MATRICES DE CONFUSIÓN
# =============================================================================

print("\n" + "="*70)
print(" MATRICES DE CONFUSIÓN")
print("="*70)

fig, axes = plt.subplots(1, 3, figsize=(20, 6))
fig.suptitle(' MATRICES DE CONFUSIÓN - COMPARATIVA DE MODELOS', fontsize=14, fontweight='bold')

for i, (nombre, metricas) in enumerate(evaluaciones.items()):
    cm = confusion_matrix(metricas['y_true'], metricas['y_pred'])

    # Normalizar
    cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

    # Mostrar
    im = axes[i].imshow(cm_norm, cmap='Blues', vmin=0, vmax=1)
    axes[i].set_title(f'{nombre}\nAccuracy: {metricas["accuracy"]*100:.1f}%', fontsize=12, fontweight='bold')
    axes[i].set_xlabel('Predicción')
    axes[i].set_ylabel('Real')

    # Colorbar
    plt.colorbar(im, ax=axes[i], fraction=0.046, pad=0.04)

plt.tight_layout()
plt.savefig('/content/12_matrices_confusion.png', dpi=150, bbox_inches='tight')
plt.show()
print(" Figura guardada: 12_matrices_confusion.png")

# =============================================================================
# CELDA 16: DEMOSTRACIÓN DE PREDICCIONES
# =============================================================================

print("\n" + "="*70)
print(" DEMOSTRACIÓN DE PREDICCIONES")
print("="*70)

# Usar el mejor modelo
modelo_final = resultados[mejor_modelo]['modelo']

# Seleccionar imágenes de prueba
fig, axes = plt.subplots(4, 4, figsize=(18, 18))
fig.suptitle(f' PREDICCIONES CON {mejor_modelo}\n(Verde = Correcto, Rojo = Incorrecto)',
             fontsize=14, fontweight='bold')

test_generator.reset()
batch_x, batch_y = next(test_generator)

predicciones = modelo_final.predict(batch_x, verbose=0)

for i in range(16):
    row, col = i // 4, i % 4

    # Imagen
    img = (batch_x[i] * 255).astype(np.uint8)
    axes[row, col].imshow(img)

    # Predicción
    pred_idx = np.argmax(predicciones[i])
    true_idx = np.argmax(batch_y[i])
    confianza = predicciones[i][pred_idx] * 100

    pred_clase = idx_to_class[pred_idx]
    true_clase = idx_to_class[true_idx]

    # Color según acierto
    color = 'green' if pred_idx == true_idx else 'red'

    titulo = f'Pred: {pred_clase[:15]}\nReal: {true_clase[:15]}\nConf: {confianza:.1f}%'
    axes[row, col].set_title(titulo, fontsize=9, color=color, fontweight='bold')
    axes[row, col].axis('off')

plt.tight_layout()
plt.savefig('/content/13_demostracion_predicciones.png', dpi=150, bbox_inches='tight')
plt.show()
print(" Figura guardada: 13_demostracion_predicciones.png")

# =============================================================================
# CELDA 17: GUARDAR MODELOS
# =============================================================================

print("\n" + "="*70)
print(" GUARDANDO MODELOS ENTRENADOS")
print("="*70)

for nombre, data in resultados.items():
    modelo = data['modelo']
    filename = f'/content/modelo_{nombre.lower().replace("-", "_")}.h5'
    modelo.save(filename)
    print(f" Guardado: {filename}")

print("\n Todos los modelos guardados correctamente")
print("="*70)

# =============================================================================
# CELDA 18: RESUMEN FINAL
# =============================================================================

print("\n" + "="*70)
print(" RESUMEN FINAL DEL PROYECTO")
print("="*70)

print("""
 SISTEMA DE IDENTIFICACIÓN DE BILLETES PERUANOS FALSIFICADOS
   Arquitectura Híbrida CNN-RNN con Procesamiento Digital de Imágenes

 DATASET:
   • Total de imágenes: {:,}
   • Categorías: {}
   • División: Train {}, Val {}, Test {}

 PROCESAMIENTO DIGITAL DE IMÁGENES (PDI):
   • {} técnicas implementadas
   • Categorías: Filtros, Contraste, Bordes, Segmentación, Morfología, Color

 MODELOS ENTRENADOS:
""".format(
    n_train + n_val + n_test,
    NUM_CLASES,
    n_train, n_val, n_test,
    len(TECNICAS_PDI)
))

for nombre, metricas in evaluaciones.items():
    print(f"   • {nombre}: {metricas['accuracy']*100:.2f}% accuracy")

print(f"""
 MEJOR MODELO: {mejor_modelo}
   • Accuracy: {evaluaciones[mejor_modelo]['accuracy']*100:.2f}%
   • F1-Score: {evaluaciones[mejor_modelo]['f1']:.4f}

 ARCHIVOS GENERADOS:
   • 01_muestras_dataset.png
   • 02_pdi_filtros_suavizado.png
   • 03_pdi_mejora_contraste.png
   • 04_pdi_deteccion_bordes.png
   • 05_pdi_segmentacion.png
   • 06_pdi_morfologia.png
   • 07_pdi_analisis_color.png
   • 08_pdi_caracteristicas.png
   • 09_comparativa_denominaciones.png
   • 10_histogramas_rgb.png
   • 11_curvas_entrenamiento.png
   • 12_matrices_confusion.png
   • 13_demostracion_predicciones.png
   • modelo_cnn.h5
   • modelo_cnn_lstm.h5
   • modelo_cnn_gru.h5
""")

print("="*70)
print(" PROYECTO COMPLETADO EXITOSAMENTE")
print("="*70)

# =============================================================================
# CELDA 19: FUNCIÓN PARA PREDECIR NUEVAS IMÁGENES
# =============================================================================

def predecir_billete(imagen_path, modelo=None):
    """
    Predice la clase de un billete dado su path o imagen.

    Args:
        imagen_path: Path a la imagen o array numpy de imagen
        modelo: Modelo a usar (por defecto el mejor modelo)

    Returns:
        Diccionario con predicción y confianza
    """
    if modelo is None:
        modelo = resultados[mejor_modelo]['modelo']

    # Cargar imagen
    if isinstance(imagen_path, str):
        img = cv2.imread(imagen_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    else:
        img = imagen_path

    # Preprocesar
    img_resized = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
    img_norm = img_resized.astype(np.float32) / 255.0
    img_batch = np.expand_dims(img_norm, axis=0)

    # Predecir
    pred = modelo.predict(img_batch, verbose=0)[0]
    pred_idx = np.argmax(pred)
    confianza = pred[pred_idx]
    clase = idx_to_class[pred_idx]

    # Mostrar resultado
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img)
    plt.title(f'Predicción: {clase}\nConfianza: {confianza*100:.1f}%',
              fontsize=12, fontweight='bold')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    colores = ['green' if i == pred_idx else 'lightgray' for i in range(len(pred))]
    plt.barh(range(len(pred)), pred, color=colores)
    plt.yticks(range(len(pred)), [idx_to_class[i][:15] for i in range(len(pred))], fontsize=8)
    plt.xlabel('Probabilidad')
    plt.title('Distribución de Probabilidades', fontsize=12, fontweight='bold')
    plt.xlim([0, 1])

    plt.tight_layout()
    plt.show()

    return {
        'clase': clase,
        'confianza': confianza,
        'probabilidades': {idx_to_class[i]: float(pred[i]) for i in range(len(pred))}
    }


print("\n FUNCIÓN DE PREDICCIÓN DISPONIBLE:")
print("   Uso: predecir_billete('/path/to/imagen.jpg')")
print("   O:   predecir_billete(imagen_array)")
```
